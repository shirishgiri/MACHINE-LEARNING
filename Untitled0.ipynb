```python
import numpy as np
import pandas as pd

np.random.seed(42)
n = 150

hours = np.random.uniform(0,40,n).round(0)
attendance = np.random.uniform(50, 100, n).round(0) # Define attendance

#A rough real-world rule,plus
#randomness - real students
#don't follow a perfect formula
score = 1.8*hours + 0.9*attendance + np.random.normal(0, 18, n) # Fixed assignment and added random noise
passed = (score > np.median(score)).astype(int)

df = pd.DataFrame({"hours_studied": hours, "attendance_pct": attendance,"passed": passed}) # Added missing comma
df.to_csv("atudents.csv",index=False)
print(df.head())
print("shape:",df.shape)
```

```python
from sklearn.model_selection \
import train_test_split
df = pd.read_csv("atudents.csv") # Corrected filename
X = df[["hours_studied",
"attendance_pct"]]
y = df["passed"]
X_train, X_test, y_train, y_test \
= train_test_split(
X, y,
test_size=0.3,
random_state=7)
print("Train:", X_train.shape)
print("Test:", X_test.shape)
```

```python
from sklearn.neighbors \
import KNeighborsClassifier
from sklearn.metrics \
import accuracy_score
model = KNeighborsClassifier(
n_neighbors=3)
model.fit(X_train, y_train)
preds = model.predict(X_test)
acc = accuracy_score(y_test, preds)
print("First 10 predictions:",
preds[:10])
print("First 10 actual: ",
y_test.values[:10])
print(f"Accuracy: {acc:.2%}")
```

```python
for k in [1, 3, 5, 10, 15]:
    m = KNeighborsClassifier(
    n_neighbors=k)
    m.fit(X_train, y_train)
    preds = m.predict(X_test)
    a = accuracy_score(
    y_test, preds)
    print(f"k={k}: "
    f"accuracy={a:.2%}")
```

```python
model = KNeighborsClassifier(
n_neighbors=3)
model.fit(X_train, y_train)
# A brand new student -- not
# in the dataset at all
new_student = pd.DataFrame([{
"hours_studied": 20.0,
"attendance_pct": 70.0}])
pred = model.predict(
new_student)
proba = model.predict_proba(
new_student)
print("Predicted:",
"Pass" if pred[0]==1
else "Fail")
print("Confidence [Fail, Pass]:",
proba.round(2))
```

### Training a Ridge Regression Model

### Checking for Overfitting with Decision Trees

Overfitting occurs when a model learns the training data too well, including its noise and specific patterns, leading to poor generalization on unseen data. We can observe this by comparing the training accuracy (how well it performs on the data it was trained on) with the test accuracy (how well it performs on new, unseen data).

Let's compare two Decision Tree models: one with lower complexity (`max_depth=2`) and another with higher complexity (`max_depth=None`, allowing it to grow fully) to see the difference in training and test performance.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Model Complexity 1: Lower Complexity (max_depth=2)
depth_low = 2
dt_low = DecisionTreeClassifier(max_depth=depth_low, random_state=1)
dt_low.fit(Xc_train, yc_train)

train_acc_low = accuracy_score(yc_train, dt_low.predict(Xc_train))
test_acc_low = accuracy_score(yc_test, dt_low.predict(Xc_test))

print(f"--- Model Complexity: max_depth={depth_low} ---")
print(f"Training Accuracy: {train_acc_low:.2%}")
print(f"Test Accuracy: {test_acc_low:.2%}")

# Model Complexity 2: Higher Complexity (max_depth=None - full tree)
depth_high = None
dt_high = DecisionTreeClassifier(max_depth=depth_high, random_state=1)
dt_high.fit(Xc_train, yc_train)

train_acc_high = accuracy_score(yc_train, dt_high.predict(Xc_train))
test_acc_high = accuracy_score(yc_test, dt_high.predict(Xc_test))

print(f"\n--- Model Complexity: max_depth={str(depth_high)} ---")
print(f"Training Accuracy: {train_acc_high:.2%}")
print(f"Test Accuracy: {test_acc_high:.2%}")

print("\nObservation: A high training accuracy and a significantly lower test accuracy for the more complex model indicates overfitting.")
```
